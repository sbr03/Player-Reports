## First Impressions
Over the several starts I've watched this season, here are my biggest takeaways:

1. Excellent Command: Detmers was consistently hitting his spots, working efficiently through counts, which allows him to 
go deep into games with a high pitch count. 

2. Weak Contact: Despite having strikeout stuff, he finds other ways to get outs to limit lengthy at-bats by generating weak contact. In his August 28th 
start against the Phillies, only 2 of the 21 Balls In Play were classified as Solid Contact according to Statcast. 

3. Curveball Usage: The curveball is not his most frequent pitch, but its usage in different counts stands out. Hitters are 
noticeably taking the pitch for a strike, a pattern the data below confirms.

## Background 
Detmers was selected 10th overall in the first round of the 2020 Draft by the Angels out of the University of Louisville, 
where he was projected to be a mid-rotation starter in scouting reports. His debut in the majors came on August 1st, 2021, allowing 
6 runs in 4.1 innings against the Athletics. He made 4 more starts that season, with a stint on the injury list sidelining him 
until September 2021. 

Detmers returned to the Angels rotation to start the 2022 campaign, throwing 129 innings in 25 starts. His 3.77 ERA and 1.21 WHIP 
on the season showed promise, highlighted by a 108-pitch no-hitter against Tampa Bay on May 10th. The following two seasons were
more difficult. Across 2023 and 2024, he struggled to find consistency from start to start. In a particularly rough stretch in 
May 2024, he allowed 26 earned runs in 24.2 innings, while never pitching through the 6th inning in any of his 5 starts that month. 
He was optioned to AAA Salt Lake in June 2024 and wasn’t recalled until September that season. His struggles continued, however, 
with back-to-back starts giving up 7 runs. 

The Angels took a different approach with Detmers in 2025. He worked exclusively out of the bullpen, appearing in 61 games with 
a 3.96 ERA. 12 of the 28 earned runs he allowed that year came within a 3-game rough stretch from April 30th to May 7th. Outside 
of that week, he was one of the Angels' most reliable relievers.

In 2026, Detmers returned to the rotation and has not looked back. He’s started a career-high 30 games this season and thrown 174 
innings to a 3.36 ERA and 199 strikeouts at the time of this writing. This past August, he threw 30 innings across 5 starts and 
allowed only 5 runs, one of the best months of his career. A 3.04 FIP and 4.33 SO/BB are also his best yet, showcasing his ability
to keep runners off the bases, which had been a struggle in previous years.

This analysis will explore his arsenal and what he’s doing well to have this resurgence in his career. The data in this
report is taken from MLB Statcast using the Python package pybaseball. This data is updated through September 14th, 2026 to include his start against the Mariners. All code used to generate these tables can be found in the ```detmers_code.ipynb``` file in this repo. 


## Pitch Location by Batter Handedness
The chart below shows the locations of Detmers’ pitches against left-handed and right-handed hitters. Two items immediately jump 
out from these charts. The first is his arsenal against lefties. The approach against LHH is very straightforward: fastballs in and

The second is his use of the curveball against RHH. There’s a ton of blue in the middle of the strike zone, which goes against what

## The "Not So Secret" Secret Weapon

As it turns out, Detmers has had some great success with his curveball. He's thrown 306 total curveballs this season, which is 
not nearly as many as his 4-seam or slider, but hitters are having trouble with it. Let’s take a look at the outcomes of Detmers 
throwing his curveball.


102 called strikes? That doesn’t seem quite right. But sure enough, the numbers don’t lie. Detmers has the highest called strike rate this season with a minimum of 300 curveballs thrown. Below are the top 10 pitchers who have the highest called strike rates on their curveball this season. We can see that even among this group of pitchers, Detmers still has the lowest total curveballs thrown and yet he gets roughly 5% more called strikes than the next pitcher.

What really sets Detmers apart is the command of his curveball. He’s thrown it for a called strike 15 more times than it has been called a ball, which is uncommon this season. He has the highest difference between called strikes and balls on curveballs this season. The table below is ranked by the highest difference, with a minimum requirement of 300 pitches thrown. 

I’d call this pitch the "Not So Secret" Secret Weapon because it truly is unique. If you watch him throw it and pay attention 
 to the hitter's reaction, you can start to understand why his called strike rate is so high. Out of the hand, it looks like 
 that ball has no business landing near the strike zone. Yet he can land it in and around the edges of the zone, which helps 
 discourage the hitter from even taking the bat off their shoulder.

He doesn’t throw the curveball often, but when he does, he throws 
it with the intention of catching the hitter off guard and giving himself an advantage in the count.

## His Best Pitch
His slider is his best all around pitch. From the movement table above, we can see that his slider has 5.52 more inches of 
downward break than a league-average left-handed slider does, which is part of what makes it so effective. In fact, it’s one of
the best strikeout pitches in the league. Below is a table of the individual pitches that have generated the most strikeouts 
so far this season, and Detmers sits at 7th with his slider, behind Chris Sale's and Chase Burns' sliders. Not bad company to be
in. To illustrate the slider in action, here's how the at-bat against Bryce Harper ended. [Harper Strikeout (Source: MLB Film Room)](https://www.mlb.com/video/bryce-harper-strikes-out-swinging-ti5hht?q=gamePk%20=%20823985%20AND%20Inning%20=%20[6]%20Order%20By%20Timestamp%20ASC&p=0&pt=Pitch%20by%20Pitch%20-%206th%20Inning )


The three tables below show the swing, whiff, and chase rates Detmers gets compared to the rest of the league on his 3 most-used pitches. 
His slider clearly stands out as his best pitch compared to the rest of the league, with a solid chase rate. What's also interesting 
to look at here is the curveball. Detmers gets below-average swing and chase percentages on the curveball, yet when the batter does swing, 
he is in the 73rd percentile for whiffs. This is yet another indicator that the curveball is tough for hitters to pick up and get a bat on.

He’s able to generate high whiff and chase rates compared to sliders across the league, and has a top Run Value on the pitch. Run Value was calculated by summing the change in run expectancy on each pitch, and Detmers' slider ranks 8th highest among all individual pitches. 

Let's take another look at what hitters are doing to the slider. Only eight extra-base hits on the pitch and seven balls in 
play that hitters made solid contact or barreled. 





The curveball and slider alone this season are a large part of Detmers' success. His slider is an excellent weapon to get weak 
contact or a whiff, while his curveball can help him steal a strike when needed. I think the next step in his development would 
be to improve his fastball offerings. There’s nothing inherently wrong with his 4-seam, but it certainly doesn’t play as well as 
his breaking pitches. If anything, it's the only pitch in his arsenal hitters are able to handle. He has a 5 Run Value on the 
4-seam (compared to 19 RV on the slider and 2 RV for the curveball), and 10 of the 18 home runs he's given up this season have been on the fastball. 

## Looking at Pitch Usage in Different Counts

The last topic to investigate is when Detmers throws his pitches and in which counts. The following 3 heatmaps show the 
initial count at which a pitch was thrown and the outcome of that pitch. For example, in a 0-0 count, Detmers has thrown his 
curveball 37 times, and the result has been a ball. The Totals by Count table is there to help illustrate how often he throws
a pitch type in a certain count.

What still jumps out is the ability to throw his curveball for strikes. Keep in mind that a strike in these tables is split
between several different outcomes, but the usage is still fascinating to me. He uses his curveball early in counts (0-0, 1-0)
to try to get ahead or even with the batter. He can then leverage that favorable count to put batters away with his slider. 
I also notice his fastball, particularly the 123 balls in a 0-0 count. Honing in on his command of the fastball early in counts
could help him get in a favorable count so that he can go to those offspeed pitches without having to worry about falling behind.


## Conclusion 
- His pitch mix is simple, yet effective. He has an unconventional use of his curveball to catch hitters off guard, instead of
 trying to get chases. The slider is one of the best strikeout pitches in baseball and seems to complement his curveball, giving
  him two elite breaking ball options in his arsenal.
- He has excellent command of his arsenal, which allows him to work through innings efficiently. He’s not afraid to throw his 
pitches for strikes and challenge hitters. 
- I am curious how he will develop further and what the next step of his evolution will be. Increasing his sinker usage to play
 off his fastball might bolster what his fastball has to offer. The sinker currently only sits at a 1% usage rate with 29 total thrown this season.

Detmers has started to show, with his aggressive pitching style and excellent command of his arsenal, that he can be a top arm 
in this league. His emergence this season has been a bright spot for the Angels, and I hope he continues to prove that their 
investment in him was worthwhile.
