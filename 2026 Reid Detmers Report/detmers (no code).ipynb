## Background 
Detmers was selected 10th overall in the first round of the 2020 Draft by the Angels out of the University of Louisville, where he was projected to be a mid-rotation starter in scouting reports. He debuted in the majors on August 1st, 2021, allowing 6 runs in 4.1 innings against the Athletics. He made 4 more starts that season, with a stint on the injury list sidelining him until September 2021. 

Detmers returned to the Angels rotation to start the 2022 campaign, throwing 129 innings in 25 starts. His 3.77 ERA and 1.21 WHIP on the season showed promise, highlighted by a 108-pitch no-hitter against Tampa Bay on May 10th. The following two seasons were more difficult. Across 2023 and 2024, he struggled to find consistency from start to start. In a particularly rough stretch in May 2024, he allowed 26 earned runs in 24.2 innings, while never pitching through the 6th inning in any of his 5 starts that month. He was optioned to AAA Salt Lake in June 2024 and wasn’t recalled until September that season. His struggles continued, however, with back-to-back starts giving up 7 runs. 

The Angels took a different approach with Detmers in 2025. He worked exclusively out of the bullpen, appearing in 61 games with a 3.96 ERA. 12 of the 28 earned runs he allowed this year came within a 3 game rough stretch from April 30th to May 7th. Outside of that week, he was one of the Angels' most reliable relievers.

In 2026, Detmers returned to the rotation and has not looked back. He’s started a career-high 29 games this season and thrown 168 innings to a 3.32 ERA and 191 strikeouts at the time of this writing. This past August, he threw 30 innings across 5 starts and allowed only 5 runs, one of the best months of his career. A 3.03 FIP and 4.24 SO/BB are also career highs, showcasing his ability to keep runners off the bases, which had been a struggle in previous years.

In this analysis, I want to explore his arsenal and what he’s doing well to have this resurgence in his career. The data in this report is taken from MLB Statcast using the Python package pybaseball. This data is updated through September 8th, 2026. I've included the code used to generate the charts and tables I created to help highlight Detmers' strengths. 


## First Impressions
Throughout the several game I have watched of Detmers this season, here were the biggest takeaways I've had. 

1. Excellent Command: Detmers was consistently hitting his spots, working efficiently through counts, which allows him to go deep into games with a high pitch count. 

2. Weak Contact: Despite having strikeout stuff, he finds other ways to get outs to limit lengthy at-bats. In his August 28th start against the Phillies, only 2 of the 21 Balls In Play were classified as Solid Contact according to Statcast. 
3. Curveball Usage: The curveball his not his most frequent pitch, but its usage in different counts stands out. Hitters are noticeably taking the pitch for a strike, a pattern the data below confirms.

## The "Not So Secret" Secret Weapon

As it turns out, Detmers has had some great success with his curveball. He's thrown 285 total curveballs this season, which is not nearly as many as his 4-seam or slider, but hitters are having trouble with it. Let’s take a look at the outcomes of Detmers throwing his curveball.


98 called strikes? That doesn’t seem quite right. But sure enough, the numbers don’t lie. Detmers' curveball is the most often called a strike in the league, by a good margin. Below are the pitchers who have the highest number of called strikes on their curveball this season. Everyone above Detmers on this list has at least 500 total curveballs thrown this season. The next closest is White Sox reliever Grant Taylor.

What really sets Detmers apart is the command of his curveball. He’s thrown it for a called strike 13 more times than it has been called a ball. Turns out that’s not so common this season either. He has the highest difference between called strikes and balls on curveballs this season. The table below is ranked by the highest difference.

I’d call this pitch the "Not So Secret" Secret Weapon because it truly is unique. If you watch him throw it and pay attention to the hitter's reaction, you can start to understand why his called strike rate is so high. Out of the hand, it looks like that ball has no business landing near the strike zone. Yet he can land it in and around the edges of the zone, which helps discourage the hitter from even taking the bat off their shoulder.

In fact, it’s one of the best strikeout pitches in the league. The three tables below show the Swing, Whiff, and Chase rates Detmers gets compared to the rest of the league on his 3 most-used pitches. His slider clearly stands out as his best pitch compared to the rest of the league, with a fantastic chase rate. What's also interesting to look at here is the curveball. Detmers gets below-average swing and Chase percentages on the curveball, yet when the batter does swing, he is in the 74th percentile for whiffs. The slider is by far his best overall pitch. He’s able to generate high whiff and chase rates compared to sliders across the league, and has a top Run Value on the pitch. Run Value was calculated by summing the change in run expectancy on each pitch, and Detmers' slider ranks 8th highest among all individual pitches. 

Let's take another look at what hitters are doing to the slider. Only seven extra-base hits on the pitch and six balls in play that hitters made solid contact or barreled. 

 



The curveball and slider alone this season are a large part of Detmers' success. His slider is an excellent weapon to get weak contact or a whiff, while his curveball can help him steal a strike when needed. I think the next step in his development would be to improve his fastball offerings. There’s nothing inherently wrong with his 4-seam, but it certainly doesn’t play as well as his breaking pitches. If anything it's the only pitch in his arsenal hitters are able to handle. He has a 5 Run Value on the 4-seam, and 10 of the 17 home runs he's given up this season have been on the fastball.

What still jumps out is the ability to throw his curveball for strikes. He uses his curveball early in counts (0-0, 1-0) to try to get ahead or even with the batter. He can then leverage that favorable count to put batters away with his slider. I also notice his fastball, particulary the 115 balls in a 0-0 count. Honing in on his command of the fastball early in counts could help him get in a favorable count so that he can go to those offspeed pitches without having to worry about falling behind.


## Conclusion 
- His pitch mix is simple, yet effective. He has an unconventional use of his curveball to catch hitters off guard, instead of trying to get chases. The slider is one of the best strikeout pitches in baseball and seems to complement his curveball, giving him 2 elite breaking ball options in his arsenal. Both pitches have a sub .200 batting average, according to Baseball Savant. 
- He has excellent command of his arsenal, which allows him to work through innings efficiently. He’s not afraid to throw his pitches for strikes and challenge hitters.
- I am curious how he will develop further and what the next step of his evolution will be. Increasing his sinker usage to play off his fastball might bolster what his fastball has to offer. 

Detmers has started to show, with his aggressive pitching style and excellent command of his arsenal, that he can be a top arm in this league. His emergence this season has been a bright spot for the Angels, and I hope he continues to prove that their investment in him was worthwhile.
